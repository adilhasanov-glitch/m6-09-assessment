```python
from ultralytics import YOLO
import pandas as pd
```

## Model improvement

```python
# Experiment 1

model = YOLO("yolo26n.pt")

results = model.train(
    data="data.yaml",
    epochs=50,
    imgsz=640,
    batch=8,

    cos_lr=True,
    patience=15,

    project="runs",
    name="exp1_coslr",

    seed=42,
    device=0
)
```

```python
# Exp 1 evaluation

model = YOLO("runs/detect/runs/exp1_coslr/weights/best.pt")

metrics = model.val(
    data="data.yaml",
    split="test"
)

print("mAP50:", metrics.box.map50)
print("mAP50-95:", metrics.box.map)
print("Precision:", metrics.box.mp)
print("Recall:", metrics.box.mr)
```

### Experiment 1 — Cosine Learning Rate + Training Stabilisation

In this experiment, I focused on improving the stability of the training process without changing the model architecture or introducing heavy augmentations.

I enabled cosine learning rate scheduling, which gradually reduces the learning rate during training instead of keeping it constant. This helps the model converge more smoothly and avoids large fluctuations in later epochs.

I also used early stopping (patience) to prevent unnecessary training once the model stops improving on the validation set. This helps reduce overfitting and saves training time.

The goal of this run was mainly to stabilise training and see whether better optimisation alone could improve performance compared to the baseline model.

```python
# Experiment 2

model = YOLO("yolo26n.pt")

results = model.train(
    data="data.yaml",
    epochs=50,
    imgsz=640,
    batch=8,

    cos_lr=True,

    mosaic=1.0,
    mixup=0.2,

    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,

    scale=0.5,
    translate=0.1,

    fliplr=0.5,

    project="runs",
    name="exp2_aug",

    seed=42,
    device=0
)
```

```python
# Exp 2 evaluation

model = YOLO("runs/detect/runs/exp2_aug/weights/best.pt")

metrics = model.val(
    data="data.yaml",
    split="test"
)

print("mAP50:", metrics.box.map50)
print("mAP50-95:", metrics.box.map)
print("Precision:", metrics.box.mp)
print("Recall:", metrics.box.mr)
```

### Experiment 2 — Improved Augmentation Strategy

For this run, I focused on improving the model’s ability to generalise to more difficult cases, especially images with multiple cats, overlapping objects, and varying lighting conditions.

To achieve this, I used stronger data augmentation during training. This included techniques such as Mosaic and MixUp, as well as colour and geometric transformations (HSV shifts, scaling, translation, and flipping). The idea was to expose the model to more diverse training samples so it becomes less sensitive to specific positions, lighting, or backgrounds.

I also kept cosine learning rate scheduling enabled to make the training process more stable over time and help the model converge more smoothly.

Overall, this experiment was aimed at improving robustness rather than just fitting the training data more closely.

## Visualization after building Docker

```python
import cv2
from pathlib import Path
import ast

PRED_PATH = "output/predictions.csv"
IMAGE_DIR = "input"
SAVE_DIR = "output/visualized"

Path(SAVE_DIR).mkdir(exist_ok=True, parents=True)

def draw_boxes(img, boxes):

    h, w = img.shape[:2]

    for box in boxes:
        try:
            x1, y1, x2, y2 = box[:4]
            conf = box[4]

            # YOLO output is in 640x640 space → scale back
            x1 = int(x1 / 640 * w)
            x2 = int(x2 / 640 * w)
            y1 = int(y1 / 640 * h)
            y2 = int(y2 / 640 * h)

            if conf < 0.25:
                continue

            cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2)
            cv2.putText(
                img,
                f"{conf:.2f}",
                (x1, y1 - 5),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5,
                (0, 255, 0),
                1
            )

        except:
            continue

    return img

with open(PRED_PATH, "r", encoding="utf-8") as f:
    lines = f.readlines()[1:]  # skip header

for line in lines:

    # 🔥 IMPORTANT: split ONLY on first comma
    image_name, boxes_raw = line.split(",", 1)

    image_name = image_name.strip()

    img_path = Path(IMAGE_DIR) / image_name
    img = cv2.imread(str(img_path))

    if img is None:
        print("Missing image:", img_path)
        continue

    try:
        boxes = ast.literal_eval(boxes_raw)
    except:
        print("Bad boxes:", image_name)
        boxes = []

    img = draw_boxes(img, boxes)

    cv2.imwrite(str(Path(SAVE_DIR) / image_name), img)

print("DONE")
```
